# Day 1 — Data Cleaning and Preprocessing
Clean a raw sales dataset using Python and Pandas.

```python
import pandas as pd
df = pd.read_csv('../data/raw_dataset.csv')
df.head()
```

```python
df.info()
df.isnull().sum()
df.duplicated().sum()
```

## Cleaning steps
- Trim whitespace and standardize text
- Normalize email casing
- Convert mixed date formats into datetime
- Convert numeric fields
- Handle missing values by column context
- Remove exact duplicate rows
- Create `Total_Sales` for validation

```python
for col in ['Customer_Name','Email','Region','Product','Status']:
    df[col] = df[col].astype('string').str.strip()
for col in ['Customer_Name','Region','Product','Status']:
    df[col] = df[col].str.title()
df['Email'] = df['Email'].str.lower()
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce', format='mixed', dayfirst=True)
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
df['Unit_Price'] = pd.to_numeric(df['Unit_Price'], errors='coerce')
df['Email'] = df['Email'].fillna('not_provided@example.com')
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
df['Unit_Price'] = df['Unit_Price'].fillna(df['Unit_Price'].median())
df = df.drop_duplicates().reset_index(drop=True)
df['Total_Sales'] = (df['Quantity'] * df['Unit_Price']).round(2)
df.to_csv('../data/cleaned_dataset.csv', index=False)
df.head()
```

```python
print('Final shape:', df.shape)
print('Remaining missing values:', int(df.isna().sum().sum()))
print('Remaining duplicate rows:', int(df.duplicated().sum()))
print(df.dtypes)
```
